# 📌 Statistical Risk in Machine Learning

In statistical learning theory, *risk* measures how much loss a prediction function produces on average.

We define it using three key components:
- **Random variable** $ Z = (X, Y) $
- **Prediction function** $ g(x) $
- **Loss function** $ L(y, g(x)) $

## ✅ True (Theoretical) Risk

$
R(g) = \int L(z, g) \, dF(z) = \mathbb{E}[L(Z, g)]
$

- $F$ is the true (unknown) data distribution (CDF)
- $L(\cdot)$ is the loss function
- The expectation is taken over all possible samples from the real world

**Interpretation:**  
> The average loss the model will produce in reality.


## 📌 Empirical Risk (Training Loss)

Since we don’t know the true distribution \(F\), we approximate using the training data:

$
R_{emp}(g) = \frac{1}{n}\sum_{i=1}^n L(z_i, g)
$

- We compute average loss on the sampled dataset
- This is what gradient descent optimizes


## 🔄 True Risk vs. Empirical Risk

| Concept | Definition | What it represents |
|--------|------------|-------------------|
| **True Risk** |  $\mathbb{E}[L(Z,g)]$  | Real-world performance |
| **Empirical Risk** | $ \frac{1}{n}\sum L(z_i,g) $ | Training performance |

Goal of learning:
> Make empirical risk as close as possible to true risk.

# 📘 Nierówność Dvoretsky–Kiefer–Wolfowitz (DKW)

Niech:
- $X_1, X_2, \ldots, X_n$ będą niezależnymi i identycznie rozłożonymi zmiennymi losowymi,
- $F^*$ oznacza prawdziwą (nieznaną) dystrybuantę.

Empiryczna dystrybuanta (EDF) definiowana jest jako:

$$
\widehat{F}_n(x) = \frac{1}{n} \sum_{i=1}^n \mathbf{1}_{[X_i,+\infty)}(x)
$$

Wyraża ona odsetek obserwacji nie większych niż $x$.

---

## 🔹 Funkcja wskaźnikowa (indicator function)

Symbol $\mathbf{1}_A(x)$ oznacza:

$$
\mathbf{1}_A(x)=
\begin{cases}
1, & \text{jeśli } x \in A, \\
0, & \text{jeśli } x \notin A
\end{cases}
$$

W kontekście EDF:

$$
\mathbf{1}_{[X_i,+\infty)}(x)=
\begin{cases}
1, & \text{jeśli } x \ge X_i, \\
0, & \text{jeśli } x < X_i
\end{cases}
$$

Funkcja ta sprawdza, czy obserwacja $X_i$ znajduje się poniżej lub na poziomie $x$.

---

# 📌 Nierówność DKW

Dla każdego $\varepsilon > 0$:

$$
P\left( \sup_x |\widehat{F}_n(x) - F^*(x)| > \varepsilon \right)
\le 2 \exp(-2n\varepsilon^2)
$$

Interpretacja:

> Prawdopodobieństwo, że empiryczna dystrybuanta oddala się od prawdziwej
> dystrybuanty o więcej niż $\varepsilon$ w dowolnym punkcie $x$, maleje
> wykładniczo wraz z wielkością próby $n$.

---

# 📌 Pasek (banda) ufności DKW

Dla poziomu ufności $1-\alpha$ ustawiamy:

$$
\varepsilon_n = \sqrt{\frac{1}{2n} \log \left( \frac{2}{\alpha} \right)}
$$

i definiujemy zakres wokół $\widehat{F}_n(x)$:

$$
\underline{C}_n(x) = \max(\widehat{F}_n(x)-\varepsilon_n, 0)
$$

$$
\overline{C}_n(x) = \min(\widehat{F}_n(x)+\varepsilon_n, 1)
$$

Wtedy:

$$
P\left(\forall x: \underline{C}_n(x) \le F^*(x) \le \overline{C}_n(x)\right)
\ge 1 - \alpha
$$

---

## 🧠 Intuicja

- $\widehat{F}_n(x)$ — proporcja danych ≤ $x$,
- funkcja wskaźnikowa liczy, czy obserwacja spełnia warunek,
- DKW gwarantuje **jednolity** (we wszystkich $x$) błąd estymacji EDF.


## 📌 Joint Distribution vs Joint PDF — najważniejsze informacje

### 🔵 Joint Distribution (łączna dystrybuanta)
- Definicja dla dwóch zmiennych losowych \(X, Y\):

$F_{X,Y}(x,y) = \mathbb{P}(X \le x,\; Y \le y)$
- Określa **prawdopodobieństwo jednoczesnego zajścia dwóch zdarzeń**.
- Zawsze istnieje — zarówno dla rozkładów dyskretnych, jak i ciągłych.
- Całkowicie opisuje wspólny rozkład zmiennych losowych.

---

### 🔴 Joint PDF (łączna gęstość prawdopodobieństwa)
- Definiowana tylko dla **ciągłych zmiennych losowych**:

$f_{X,Y}(x,y)$
- Prawdopodobieństwo w obszarze \(A\):

$\mathbb{P}((X,Y)\in A)=\iint_A f_{X,Y}(x,y)\,dx\,dy$
- Powiązana z dystrybuantą:

$f_{X,Y}(x,y)=\frac{\partial^2}{\partial x\,\partial y}F_{X,Y}(x,y)$
- Normalizacja:

$\iint_{\mathbb{R}^2} f_{X,Y}(x,y)\,dx\,dy = 1$

---

### 🔸 Związek między CDF i PDF
- Dystrybuanta powstaje przez całkowanie gęstości:

$F_{X,Y}(x,y)=\int_{-\infty}^{x}\int_{-\infty}^{y}f_{X,Y}(u,v)\,dv\,du$

---

### 🔹 Dla zmiennych dyskretnych
- Zamiast PDF stosuje się **joint PMF**:

$p_{X,Y}(x,y)=\mathbb{P}(X=x,\;Y=y)$

---

### ✅ Kluczowe punkty
- **Joint CDF → prawdopodobieństwo zdarzeń łącznych.**
- **Joint PDF → gęstość, którą trzeba integrować, aby uzyskać prawdopodobieństwo.**
- PDF ≠ prawdopodobieństwo punktowe.
- PMF jest odpowiednikiem joint PDF dla rozkładów dyskretnych.


# Conditions for Random Variables (RVs) to be Independent

Two random variables (RVs) are **independent** if knowing the value of one gives no information about the other. Formally, the conditions can be stated as follows:

---

## 1. Using Cumulative Distribution Functions (CDFs)

Let $F_X(x)$ and $F_Y(y)$ be the marginal CDFs of $X$ and $Y$, and $F_{X,Y}(x,y)$ be their joint CDF. Then:

$X \text{ and } Y \text{ are independent } \iff F_{X,Y}(x, y) = F_X(x) \cdot F_Y(y), \quad \forall x, y$

---

## 2. Using Probability Density Functions (PDFs) — Continuous Case

For continuous RVs with joint PDF $f_{X,Y}(x,y)$ and marginal PDFs $f_X(x)$ and $f_Y(y)$:

$X \text{ and } Y \text{ are independent } \iff f_{X,Y}(x, y) = f_X(x) \cdot f_Y(y), \quad \forall x, y$

---

## 3. Using Probability Mass Functions (PMFs) — Discrete Case

For discrete RVs with joint PMF $p_{X,Y}(x,y)$ and marginal PMFs $p_X(x)$ and $p_Y(y)$:

$X \text{ and } Y \text{ are independent } \iff p_{X,Y}(x, y) = p_X(x) \cdot p_Y(y), \quad \forall x, y$

---

### Summary

- **CDF condition:** joint CDF factorizes into marginals  
- **PDF condition (continuous):** joint PDF factorizes into marginals  
- **PMF condition (discrete):** joint PMF factorizes into marginals  

Independence can be checked by verifying any of these conditions depending on whether the variables are discrete or continuous.


# 📈 Key Probability Inequalities

These inequalities provide **bounds** on the probability of a random variable deviating from its mean, often useful when the full distribution is unknown.

---

## 1. Markov's Inequality (Bounding a Non-Negative Variable)

* **Applies to:** Any **non-negative** random variable $X$ ($P(X \ge 0) = 1$).
* **Formula:**
    $$P(X \ge a) \le \frac{E[X]}{a}$$
    * $E[X]$ is the **expected value** (mean) of $X$.

---

## 2. Chebyshev's Inequality (Bounding Deviation from the Mean)

* **Applies to:** Any random variable $X$ with a defined **mean** ($\mu$) and **finite variance** ($\sigma^2$).

### Formula 2.1: Absolute Deviation (Tail Probability)
* **Purpose:** Provides the maximum probability of falling **outside** the mean $\pm$ distance $a$.
* **Formula:**
    $$P(|X - \mu| \ge a) \le \frac{\sigma^2}{a^2}$$

### Formula 2.2: Standard Deviations (Central Probability)
* **Purpose:** Provides the minimum probability of falling **inside** the mean $\pm$ $k$ standard deviations.
* **Formula:**
    $$P(|X - \mu| < k\sigma) \ge 1 - \frac{1}{k^2}$$

### Formula 2.3: Higher Moment Generalization
* **Purpose:** Uses any finite $p$-th central moment.
* **Formula:**
    $$P(|X - \mu| \ge a) \le \frac{E[|X - \mu|^p]}{a^p}$$

---

## 3. Hoeffding's Inequality (Bounding Sums of Bounded Random Variables)

* **Applies to:** A sum (or average) of **independent** random variables $X_1, X_2, \dots, X_n$, where each $X_i$ is **bounded** ($X_i \in [a, b]$).

### Formula 3.1: One-Sided Bound (Specific Deviation)
* **Purpose:** Provides an exponential upper bound on the probability that the sample mean $\bar{X}_n$ is **greater than** its expected value by at least $\epsilon$.
* **Formula:**
    $$\mathbb{P}(\bar{X}_n - \mathbb{E}[\bar{X}_n] \ge \epsilon) \le e^{-\frac{2n\epsilon^2}{(b-a)^2}}$$
    * This assumes $X_i$ are i.i.d. and bounded by $[a, b]$.

### Formula 3.2: Two-Sided Bound (General Deviation)
* **Purpose:** Provides an exponential upper bound on the probability of deviation in **either direction** (above or below the mean).
* **Formula:**
    $$P(|\bar{X}_n - E[\bar{X}_n]| \ge \epsilon) \le 2 \exp\left(-\frac{2n\epsilon^2}{(b - a)^2}\right)$$

# How to Construct a Confidence Interval Using a Concentration Inequality

Given i.i.d. random variables $X_1, \dots, X_n$ with mean $\mu$, define the sample mean:
$\overline{X}_n = \frac{1}{n} \sum_{i=1}^n X_i$.

A concentration inequality gives a bound of the form
$P(|\overline{X}_n - \mu| > \varepsilon) \le \delta(\varepsilon, n)$.
To convert this into a confidence interval, use the complement:

$$
P(|\overline{X}_n - \mu| \le \varepsilon) = 1 - P(|\overline{X}_n - \mu| > \varepsilon) \ge 1 - \delta.
$$

Rewriting the event $|\overline{X}_n - \mu| \le \varepsilon$ gives the confidence statement:
$$
P\big(\mu \in [\,\overline{X}_n - \varepsilon,\; \overline{X}_n + \varepsilon\,]\big) \ge 1 - \delta.
$$

So the recipe is: find $\varepsilon$ (depending on $n,\delta$) from the inequality $P(|\overline{X}_n - \mu| > \varepsilon) \le \delta$, then report the $(1-\delta)$ confidence interval $\overline{X}_n \pm \varepsilon$.

---

## **Step 1 — Example inequalities**

- **Chebyshev:**  
  $P(|\overline{X}_n - \mu| > \varepsilon) \le \frac{\sigma^2}{n\varepsilon^2}$.

- **Hoeffding (bounded $X_i \in [a,b]$):**  
  $P(|\overline{X}_n - \mu| > \varepsilon) \le 2 \exp\!\left(-\frac{2n\varepsilon^2}{(b-a)^2}\right)$.

---

## **Step 2 — Solve for $\varepsilon$ given $\delta$**

Set the RHS equal to $\delta$ and solve for $\varepsilon$:

- Hoeffding:
  $2 e^{-2n\varepsilon^2/(b-a)^2} = \delta
  \Rightarrow
  \varepsilon = \frac{b-a}{\sqrt{2n}} \sqrt{\ln\!\left(\frac{2}{\delta}\right)}.$

- Chebyshev:
  $\frac{\sigma^2}{n\varepsilon^2} = \delta
  \Rightarrow
  \varepsilon = \sqrt{\frac{\sigma^2}{n\delta}}.$

---

## **Step 3 — Confidence interval**

A $(1-\delta)$ confidence interval for $\mu$ is:
$$
\mu \in [\,\overline{X}_n - \varepsilon,\; \overline{X}_n + \varepsilon\,],
$$
where $\varepsilon$ is obtained from the chosen concentration inequality.

---

## **Summary**
1. Pick inequality.  
2. Set tail bound = $\delta$.  
3. Solve for $\varepsilon$.  
4. Report $\overline{X}_n \pm \varepsilon$, which satisfies
   $P\big(\mu \in [\overline{X}_n - \varepsilon,\; \overline{X}_n + \varepsilon]\big) \ge 1-\delta$.


## Nierówność Hoeffdinga

Niech $Y$ będzie zmienną losową taką, że  
$Y \in [a,b] \quad \text{p.n.}$

Wówczas dla dowolnego $\varepsilon > 0$ zachodzi:  
$\mathbb{P}\big( |Y - \mathbb{E}Y| \ge \varepsilon \big)
\;\le\;
2 \exp\!\left( -\frac{2\varepsilon^2}{(b-a)^2} \right)$

Brakuje $n$ we wzorze ponieważ $Y$ nie jest średnią.

---

## Wybór $\varepsilon$ dla poziomu ufności $1-\delta$

Aby otrzymać poziom ufności $1-\delta$, wybieramy $\varepsilon$ tak, że:  
$2 \exp\!\left( -\frac{2\varepsilon^2}{(b-a)^2} \right) = \delta$

co daje:  
$\varepsilon = (b-a)\sqrt{\frac{1}{2}\ln\!\left(\frac{2}{\delta}\right)}$

---

## Przedział ufności

Z prawdopodobieństwem co najmniej $1-\delta$ zachodzi:  
$$
Y - \varepsilon,
\;
Y + \varepsilon
$$

czyli przedział ufności dla $\mathbb{E}Y$ ma postać:  
$
Y - (b-a)\sqrt{\tfrac{1}{2}\ln(\tfrac{2}{\delta})},
\;
Y + (b-a)\sqrt{\tfrac{1}{2}\ln(\tfrac{2}{\delta})}
$


### **Law of Large Numbers (LLN)**

The **Law of Large Numbers** describes how sample averages behave as the sample size grows.

**Setup:**  
Let \(X_1, X_2, \ldots\) be i.i.d. random variables with finite mean $\mu = \mathbb{E}[X_i]$.  
Define the sample mean:  
$$
\overline{X}_n = \frac{1}{n}\sum_{i=1}^{n} X_i .
$$

**Weak Law of Large Numbers (WLLN):**  
$$
\overline{X}_n \xrightarrow{P} \mu
$$
The sample mean converges *in probability* to the true mean.

**Strong Law of Large Numbers (SLLN):**  
$$
\overline{X}_n \xrightarrow{\text{a.s.}} \mu
$$
The sample mean converges *almost surely* to the true mean.

**Intuition:**  
As $n$ gets large, random fluctuations average out, and $\overline{X}_n$ becomes very close to $\mu$.

---

### **Central Limit Theorem (CLT)**

The **Central Limit Theorem** describes the distribution of the sample mean for large samples.

**Setup:**  
Same as before: $X_1, X_2, \ldots$ i.i.d. with mean $\mu$ and variance $\sigma^2 < \infty$.

**Statement:**  
$$
\frac{\overline{X}_n - \mu}{\sigma / \sqrt{n}} \xrightarrow{d} \mathcal{N}(0,1)
$$

**Meaning:**  
Regardless of the original distribution of the $X_i$, the standardized sample mean becomes approximately normal for large $n$.

**Intuition:**  
Totals or averages of many small independent contributions tend to follow a bell curve — even when the underlying data are not normal.

---

### **LLN vs. CLT**

- **LLN:** Ensures $\overline{X}_n$ gets close to $\mu$.  
- **CLT:** Describes *how* $\overline{X}_n$ fluctuates around $\mu$ for large $n$.  
- LLN gives **convergence**, CLT gives **distributional shape**.



## 📊 Modes of Stochastic Convergence: Probability vs. Quadratic Mean ($\mathbf{L^2}$)

This table contrasts two fundamental ways a sequence of Random Variables (RVs), $\{X_n\}$, can mathematically converge to a limit RV, $X$. 

| Mode of Convergence | Notation | Definition | Intuition & Significance |
| :--- | :--- | :--- | :--- |
| **Convergence in Probability** | $X_n \xrightarrow{P} X$ | $\lim_{n \to \infty} P(|X_n - X| \geq \epsilon) = 0 \quad \text{for all } \epsilon > 0$ | **Weakest of the two.** Focuses on the **frequency** of large errors. It means the chance of a significant deviation ($\epsilon$) vanishes as $n \to \infty$. This underlies the **Weak Law of Large Numbers (WLLN)** and defines **consistent estimators** in statistics. |
| **Convergence in Quadratic Mean** (or Mean Square, $L^2$) | $X_n \xrightarrow{q.m.} X$ or $X_n \xrightarrow{L^2} X$ | $$\lim_{n \to \infty} E[(X_n - X)^2] = 0$$ | **Stronger than convergence in probability.** Focuses on the average magnitude of the **squared error** (the Mean Squared Error, or MSE). It heavily penalizes large, rare errors (high variance) and requires that both the mean and variance of $X_n$ stabilize. |

---

### 🔑 Key Relationship

The most important relationship is the hierarchy:

$$\text{Convergence in Quadratic Mean } \implies \text{ Convergence in Probability}$$

* $L^2$ convergence is a **sufficient condition** for convergence in probability.
* The reverse implication is **not true**.

# Short Guide: Using the Delta Method to Find a Limiting Distribution

## 1. Purpose of the Delta Method
The **delta method** is used to derive the **asymptotic (limiting) distribution** of a smooth function of an estimator when the estimator itself has a known limiting distribution, typically normal.

It is most commonly applied when:
- You know the asymptotic distribution of an estimator $\hat{\theta}_n$
- You want the asymptotic distribution of $g(\hat{\theta}_n)$ for some smooth function $g$

---

## 2. Basic Setting
Assume:
- $\hat{\theta}_n \xrightarrow{p} \theta$
- $\sqrt{n}(\hat{\theta}_n - \theta) \xrightarrow{d} N(0, \sigma^2)$
- $g(\cdot)$ is differentiable at $\theta$

---

## 3. First-Order Delta Method
Use a first-order Taylor expansion of $g(\hat{\theta}_n)$ around $\theta$:

$g(\hat{\theta}_n) \approx g(\theta) + g'(\theta)(\hat{\theta}_n - \theta)$

Multiply both sides by $\sqrt{n}$:

$\sqrt{n}(g(\hat{\theta}_n) - g(\theta)) \approx g'(\theta)\sqrt{n}(\hat{\theta}_n - \theta)$

Taking limits in distribution:

$\sqrt{n}(g(\hat{\theta}_n) - g(\theta)) \xrightarrow{d} N(0, [g'(\theta)]^2 \sigma^2)$

---

## 4. Final Result
If $\sqrt{n}(\hat{\theta}_n - \theta) \xrightarrow{d} N(0, \sigma^2)$, then:

$\sqrt{n}(g(\hat{\theta}_n) - g(\theta)) \xrightarrow{d} N(0, [g'(\theta)]^2 \sigma^2)$

---

## 5. Multivariate Version (Brief)
If:
- $\sqrt{n}(\hat{\theta}_n - \theta) \xrightarrow{d} N(0, \Sigma)$
- $g: \mathbb{R}^k \to \mathbb{R}$ is differentiable

Then:

$\sqrt{n}(g(\hat{\theta}_n) - g(\theta)) \xrightarrow{d} N(0, \nabla g(\theta)^\top \Sigma \nabla g(\theta))$

---

## 6. When First-Order Fails
If $g'(\theta) = 0$, use a **second-order delta method**, based on the second derivative. The convergence rate may change (e.g., $n$ instead of $\sqrt{n}$).

---

## 7. Common Pitfalls
- Forgetting to check differentiability of $g$
- Evaluating derivatives at the wrong point
- Ignoring cases where the first derivative is zero
- Confusing convergence in probability with convergence in distribution

---

## 8. Typical Workflow
1. Identify $\hat{\theta}_n$ and its limiting distribution
2. Verify consistency: $\hat{\theta}_n \xrightarrow{p} \theta$
3. Choose function $g$
4. Compute $g'(\theta)$ (or gradient)
5. Apply the delta method formula


# Linear Congruential Generator (LCG)

The **Linear Congruential Generator (LCG)** is a deterministic algorithm for producing a sequence of numbers that approximates samples from the uniform distribution on the interval (0, 1). It is defined by a simple linear recurrence relation combined with modular arithmetic.

---

## 1. Definition

An LCG generates a sequence of integers $\{X_n\}$ according to

$X_{n+1} = (a X_n + c) \bmod m$

where:
- $m > 0$ is the **modulus**,
- $a$ is the **multiplier**,
- $c$ is the **increment**,
- $X_0$ is the **seed** (initial value).

Each $X_n$ takes values in the set $\{0, 1, \dots, m-1\}$.

---

## 2. Producing Uniform(0,1) Values

To obtain numbers in the interval $(0,1)$, the integer output is normalized:

$U_n = \dfrac{X_n}{m}$

The sequence $\{U_n\}$ is intended to behave like independent realizations from $U(0,1)$.

---

## 3. Step-by-Step Algorithm

1. Choose parameters $m$, $a$, $c$, and an initial seed $X_0$.
2. Compute $X_{n+1} = (a X_n + c) \bmod m$.
3. Normalize: $U_{n+1} = X_{n+1} / m$.
4. Repeat steps 2–3 to generate as many values as needed.

The procedure is fully deterministic once the parameters and seed are fixed.

---

## 4. Interpretation of the Parameters

- **Modulus $m$**  
  Determines the range of integer values and the maximum possible period of the sequence.

- **Multiplier $a$**  
  Controls how the previous state influences the next one. Its choice strongly affects the quality of the sequence.

- **Increment $c$**  
  Shifts the sequence and helps avoid trivial cycles (e.g., getting stuck at zero).

- **Seed $X_0$**  
  Sets the starting point of the sequence. Different seeds produce different sequences with the same parameters.

---

## 5. Period and Cycling

Because the generator uses modular arithmetic, the sequence must eventually repeat.  
The **period** is the number of generated values before the sequence starts repeating.

The maximum possible period is $m$. It is achieved if and only if the following conditions hold:

1. $\gcd(c, m) = 1$  
2. $a - 1$ is divisible by all prime factors of $m$  
3. If $m$ is a multiple of 4, then $a - 1$ is also a multiple of 4

When these conditions are satisfied, the generator cycles through all values $\{0, \dots, m-1\}$ exactly once per period.

---

## 6. Deterministic Nature

Although LCG outputs appear random, the sequence is completely deterministic:
- given $X_n$, all future values are fixed,
- repeating the same seed and parameters reproduces the same sequence.

This determinism is why LCG outputs are called *pseudorandom* rather than truly random.

---

## 7. Conceptual Summary

An LCG:
- evolves a state using a linear function modulo $m$,
- produces integer values that are scaled to $(0,1)$,
- generates a finite, repeating sequence,
- relies critically on careful parameter selection for acceptable behavior.

In essence, an LCG is a simple mathematical mechanism that converts arithmetic recurrence into a stream of numbers that approximate realizations from $U(0,1)$.


# Markov Chains

A **Markov chain** is a stochastic process that models a sequence of random states evolving over time, where the future depends only on the present state and not on the past history.

---

## 1. Definition

Let $\{X_n\}_{n \ge 0}$ be a sequence of random variables taking values in a **state space** $S$ (finite or countable).

The process is a **Markov chain** if it satisfies the **Markov property**:

$P(X_{n+1} = j \mid X_n = i, X_{n-1}, \dots, X_0)
= P(X_{n+1} = j \mid X_n = i)$

for all states $i, j \in S$ and all $n$.

---

## 2. Transition Probabilities

The dynamics of the chain are described by **transition probabilities**:

$p_{ij} = P(X_{n+1} = j \mid X_n = i)$

These probabilities satisfy:
- $p_{ij} \ge 0$
- $\sum_{j \in S} p_{ij} = 1$ for each $i$

---

## 3. Transition Matrix

For a finite state space, the transition probabilities form a **transition matrix**:

$P =
\begin{pmatrix}
p_{11} & p_{12} & \cdots \\
p_{21} & p_{22} & \cdots \\
\vdots & \vdots & \ddots
\end{pmatrix}$

Properties:
- each row sums to 1,
- $P^n$ gives $n$-step transition probabilities.

---

## 4. $n$-Step Transition Probabilities

The probability of moving from state $i$ to state $j$ in $n$ steps is:

$p_{ij}^{(n)} = P(X_n = j \mid X_0 = i)$

Matrix form:
$P^{(n)} = P^n$

---

## 5. State Classification

### Accessibility
State $j$ is **accessible** from $i$ if:
$p_{ij}^{(n)} > 0$ for some $n \ge 1$

---

### Communication
States $i$ and $j$ **communicate** if:
- $i$ is accessible from $j$, and
- $j$ is accessible from $i$

Communication is an equivalence relation that partitions the state space.

---

### Irreducibility
A Markov chain is **irreducible** if all states communicate with each other, i.e., the state space consists of a single communicating class.

The chain is irreducible if we have **transient state** (state that we leave at some point and never come back).
**Recurrent state** - other states than transient states.

---

## 6. Periodicity

The **period** of a state $i$ is:

$d(i) = \gcd \{ n \ge 1 : p_{ii}^{(n)} > 0 \}$ ==> to jest NWD wszystkim możliwych scieżek powrotu do stanu. Czyli np. jeśli mam dwie ścieżki z liczbą kroków 2 i 3 to d=1.

- If $d(i) = 1$, the state is **aperiodic**
- If $d(i) > 1$, the state is **periodic**

In an irreducible chain, all states have the same period.

A markov chain is irreducible ==> all states are aperiodic or all states have the same period.

So if markov chain is irreducible and has a loop on some state then the chain is aperiodic.

---

## 7. Recurrence and Transience

Let:
$f_{ii} = P(\text{return to state } i \text{ at some time } n \ge 1)$

- State $i$ is **recurrent** if $f_{ii} = 1$ --> na pewno kiedyś wrócimy do tego stanu
- State $i$ is **transient** if $f_{ii} < 1$ --> kiedyś bezpowrotnie wyjdziemy z tego stanu i nie wrócimy do niego.

In a recurrent state, the chain returns infinitely often with probability 1.

---

## 8. Positive and Null Recurrence

For a recurrent state $i$, define the **expected return time**:

$E_i[T_i] = E(\text{time to first return to } i)$

- **Positive recurrent**: $E_i[T_i] < \infty$
- **Null recurrent**: $E_i[T_i] = \infty$

For finite irreducible chains, all states are positive recurrent.

---

## 9. Stationary Distribution

A probability distribution $\pi = (\pi_i)_{i \in S}$ is **stationary** if:

$\pi_j = \sum_{i \in S} \pi_i p_{ij}$

Matrix form:
$\pi = \pi P$

Interpretation:
- if $X_0 \sim \pi$, then $X_n \sim \pi$ for all $n$.

(Existence of Stationary distribution). For any irreducible and aperiodic Markov chain there exists at least one stationary distribution.

---

## 10. Existence and Uniqueness of Stationary Distribution

- An irreducible, positive recurrent chain has a stationary distribution.
- If the chain is also aperiodic, the stationary distribution is unique.

---

## 11. Limiting Distribution

A distribution $\pi$ is a **limiting distribution** if:

$\lim_{n \to \infty} P(X_n = j \mid X_0 = i) = \pi_j$

for all $i, j$.

For a finite-state Markov chain:
- irreducible,
- aperiodic,

the limiting distribution exists and equals the stationary distribution.

---

## 12. Detailed Balance and Reversibility

A distribution $\pi$ satisfies **detailed balance** if:

$\pi_i p_{ij} = \pi_j p_{ji}$

If detailed balance holds, the chain is **reversible** and $\pi$ is stationary.

---

## 13. Ergodicity

A Markov chain is **ergodic** if it is:
- irreducible,
- aperiodic,
- positive recurrent.

Ergodicity implies:
- convergence to the stationary distribution,
- long-run time averages equal expectations under $\pi$.

---

## 14. Long-Run Frequencies

For an ergodic chain:

$\frac{1}{n} \sum_{k=1}^n \mathbf{1}_{\{X_k = i\}} \to \pi_i \quad \text{a.s.}$

This means $\pi_i$ equals the long-run proportion of time spent in state $i$.

---

## 15. Conceptual Summary

A Markov chain:
- evolves according to fixed transition probabilities,
- forgets its past beyond the current state,
- exhibits long-run behavior characterized by recurrence and stationarity,
- converges to a unique stationary distribution under mild conditions.

Understanding concepts such as irreducibility, periodicity, and stationarity is essential for analyzing both short-term dynamics and long-term behavior of Markov chains.


# Nierówność Bennetta i przedział ufności

## Nierówność Bennetta

Niech $X_1,\dots,X_n$ będą niezależnymi zmiennymi losowymi takimi, że:

- $\mathbb{E}[X_i] = 0$
- $X_i \le b$ prawie na pewno
- $\operatorname{Var}(X_i) = \sigma_i^2$

Definiujemy łączną wariancję:
$\sigma^2 = \sum_{i=1}^n \sigma_i^2$

Wówczas dla każdego $t > 0$ zachodzi nierówność Bennetta:
$$
\mathbb{P}\left( \sum_{i=1}^n X_i \ge t \right)
\le
\exp\!\left(
-\frac{\sigma^2}{b^2}
\left[(1+u)\ln(1+u) - u\right]
\right),
$$
gdzie
$$
u = \frac{b\,t}{\sigma^2}.
$$

---

## Zastosowanie do średniej empirycznej

Niech $Y_1,\dots,Y_n$ będą zmiennymi i.i.d. takimi, że:

- $Y_i \in [a,b]$
- $\mathbb{E}[Y_i] = \mu$
- $\operatorname{Var}(Y_i) \le \sigma^2$

Definiujemy średnią empiryczną:
$$
\bar Y = \frac{1}{n}\sum_{i=1}^n Y_i.
$$

Po scentralizowaniu $X_i = Y_i - \mu$ otrzymujemy:
$$
\mathbb{P}(\bar Y - \mu \ge \varepsilon)
\le
\exp\!\left(
-\frac{n\sigma^2}{(b-a)^2}
h\!\left(
\frac{(b-a)\varepsilon}{\sigma^2}
\right)
\right),
$$
gdzie
$$
h(u) = (1+u)\ln(1+u) - u.
$$

---

## Promień przedziału ufności (postać jawna – Bernstein)

Odwracając powyższą nierówność (w praktyce korzystając z jej uproszczenia Bernsteina),
otrzymujemy następujący promień przedziału ufności:

$$
\varepsilon
=
\sqrt{
\frac{2\sigma^2 \ln(2/\delta)}{n}
}
+
\frac{(b-a)\ln(2/\delta)}{3n}.
$$

Z prawdopodobieństwem co najmniej $1-\delta$ zachodzi:
$$
|\bar Y - \mu| \le \varepsilon.
$$

---

## Znaczenie symboli

- $n$ – liczba obserwacji (rozmiar próby)
- $Y_i$ – pojedyncza obserwacja zmiennej losowej
- $\bar Y$ – średnia empiryczna z próby
- $\mu$ – prawdziwa wartość oczekiwana zmiennej losowej
- $a,b$ – dolne i górne ograniczenie zmiennej losowej ($Y_i \in [a,b]$)
- $\sigma^2$ – górne ograniczenie wariancji pojedynczej obserwacji
- $\delta$ – poziom istotności (prawdopodobieństwo błędu)
- $\varepsilon$ – promień przedziału ufności
- $h(u)$ – funkcja Bennetta opisująca tempo koncentracji
- $t$ – poziom odchylenia sumy od wartości oczekiwanej

---

## Interpretacja

Nierówność Bennetta daje wykładniczne ograniczenie na prawdopodobieństwo dużych
odchyleń sumy (lub średniej) zmiennych losowych, uwzględniając zarówno ich wariancję,
jak i ograniczenie wartości. Promień $\varepsilon$ określa, jak daleko średnia
empiryczna może się znajdować od prawdziwej średniej z zadanym poziomem ufności.


### Two Common Definitions of the Likelihood Function

**1. Conditional (Discriminative) Likelihood**

$\mathcal{L}(\theta) = \prod_{i=1}^n p(y_i \mid x_i; \theta)$

- Used in **discriminative models** (e.g. logistic regression).
- Models the probability of the label given the input.
- Focuses directly on predicting $y$ from $x$.
- Does **not** model how the data $x$ is generated.

---

**2. Joint / Generative Likelihood**

$\mathcal{L}(\theta) = \prod_{i=1}^n p(x_i, y_i \mid \theta)
= \prod_{i=1}^n p(x_i \mid y_i; \theta)\, p(y_i)$

- Used in **generative models** (e.g. LDA, QDA, Naive Bayes).
- Models how the data is generated for each class.
- Allows computing $p(y \mid x)$ via Bayes’ rule.

---

# Moment Generating Function (MGF)

## 1. Definicja

Niech $X$ będzie zmienną losową.  
**Moment generating function (MGF)** zmiennej $X$ jest zdefiniowana jako
$$
M_X(t) = \mathbb E[e^{tX}],
$$
dla tych $t \in \mathbb R$, dla których wartość oczekiwana jest skończona.

Często rozważa się wersję scentrowaną:
$$
\mathbb E[e^{t(X-\mathbb EX)}].
$$

---

## 2. Po co jest MGF

MGF służy do:
- generowania momentów,
- badania istnienia momentów,
- analizy ogonów rozkładu,
- dowodów nierówności koncentracji miary,
- teorii dużych odchyleń.

Intuicja:  
MGF mierzy, jak szybko rośnie $e^{tX}$, co bezpośrednio kontroluje ciężkość ogonów rozkładu.

---

## 3. Generowanie momentów

Jeżeli MGF istnieje w otoczeniu zera, to:
$$
\mathbb E[X^k] = M_X^{(k)}(0).
$$

### Przykłady
- Pierwszy moment:
$$
M_X'(0) = \mathbb EX
$$
- Drugi moment:
$$
M_X''(0) = \mathbb E[X^2]
$$
- Wariancja:
$$
\mathrm{Var}(X) = M_X''(0) - (M_X'(0))^2
$$

Wniosek:  
Istnienie MGF w otoczeniu zera implikuje istnienie wszystkich momentów.

---

## 4. MGF a ogony rozkładu

MGF pozwala ograniczać prawdopodobieństwo dużych odchyleń.

### Nierówność Chernoffa
Dla dowolnego $t>0$:
$$
\mathbb P(X \ge \varepsilon)
\le
e^{-t\varepsilon} \mathbb E[e^{tX}]
=
e^{-t\varepsilon} M_X(t).
$$

Jeżeli potrafimy ograniczyć MGF, otrzymujemy wykładnicze ogony.

---

## 5. MGF a koncentracja miary

### Zmienne sub-Gaussian
$X$ jest sub-Gaussian, jeśli istnieje $\sigma>0$ takie że
$$
\mathbb E[e^{t(X-\mathbb EX)}]
\le
\exp\left(\frac{\sigma^2 t^2}{2}\right)
\quad \forall t \in \mathbb R.
$$

Implikacje:
- ogony typu $e^{-c x^2}$,
- koncentracja średniej rzędu $e^{-n\varepsilon^2}$.

---

### Zmienne sub-Exponential
$X$ jest sub-Exponential, jeśli
$$
\mathbb E[e^{t(X-\mathbb EX)}]
\le
\exp\left(\frac{\nu^2 t^2}{2}\right)
\quad \text{dla } |t| < 1/b.
$$

Implikacje:
- ogony typu $e^{-c x}$,
- koncentracja typu Bernstein (dwa reżimy).

---

## 6. Brak MGF

Jeżeli istnieje tylko skończona wariancja:
$$
\mathbb E[X^2] < \infty,
$$
to:
- MGF może nie istnieć dla żadnego $t \neq 0$,
- brak wykładnicznej koncentracji,
- pozostaje nierówność Czebyszewa:
$$
\mathbb P(|X-\mathbb EX| \ge \varepsilon)
\le
\frac{\mathrm{Var}(X)}{\varepsilon^2}.
$$

---

## 7. Podsumowanie

- MGF generuje momenty przez różniczkowanie w $0$.
- Kontrola MGF implikuje kontrolę ogonów.
- Wykładniczna koncentracja wymaga istnienia MGF w otoczeniu zera.

# Negative Log-Likelihood (NLL) Function

## 1. Definition

The negative log-likelihood (NLL) is a loss function commonly used in probabilistic models.
Given a dataset of $n$ independent observations $(x_1, x_2, ..., x_n)$
and a parametric model $p(x \mid \theta)$ with parameters $\theta$, the likelihood is:

$$
L(\theta) = \prod_{i=1}^n p(x_i \mid \theta)
$$

Taking the logarithm gives the log-likelihood:

$$
\ell(\theta) = \log L(\theta) = \sum_{i=1}^n \log p(x_i \mid \theta)
$$

The negative log-likelihood is simply:

$$
NLL(\theta) = - \ell(\theta) = - \sum_{i=1}^n \log p(x_i \mid \theta)
$$

This is the function that we aim to **minimize** to estimate the parameters $\theta$.

---
## 2. Step-by-Step Process

### Step 1: Define the model probability
Choose a probability distribution for your data.
For example, for Bernoulli-distributed data $y_i \in \{0,1\}$ with success probability $p$:

$$
p(y_i \mid p) = p^{y_i} (1-p)^{1-y_i}
$$

---
### Step 2: Write the log-likelihood
For Bernoulli data, the log-likelihood is:

$$
\ell(p) = \sum_{i=1}^n \left[ y_i \log p + (1-y_i) \log (1-p) \right]
$$

---
### Step 3: Negate the log-likelihood

$$
NLL(p) = - \sum_{i=1}^n \left[ y_i \log p + (1-y_i) \log (1-p) \right]
$$

This is the exact function we minimize in order to estimate the parameter $p$.

---
### Step 4: Implementation in Python

```python
import numpy as np

def bernoulli_nll(p, y):
    # Clip p to avoid log(0)
    p = np.clip(p, 1e-10, 1 - 1e-10)
    return -np.sum(y * np.log(p) + (1 - y) * np.log(1 - p))
```

- `y` is a numpy array of 0s and 1s.
- `p` is the probability of success.

---
### Step 5: Parameter estimation via optimization

```python
from scipy.optimize import minimize

y_data = np.array([1, 0, 1, 1, 0])

result = minimize(lambda p: bernoulli_nll(p, y_data), x0=[0.5], bounds=[(0,1)])
estimated_p = result.x[0]
print("Estimated probability:", estimated_p)
```

---
## Notes

- The same approach generalizes to other distributions (Gaussian, Poisson, categorical, etc.).
- Negating the log-likelihood is essential because most optimization libraries **minimize** functions.

# Vapnik–Chervonenkis (VC) Dimension

## 1. Definition

The Vapnik–Chervonenkis (VC) dimension is a measure of the **capacity** or
**complexity** of a class of classifiers.

Let $\mathcal{H}$ be a hypothesis class, i.e. a set of binary classifiers
$h : \mathcal{X} \rightarrow \{0,1\}$.

A set of points $\{x_1, \dots, x_n\} \subset \mathcal{X}$ is said to be
**shattered** by $\mathcal{H}$ if for every possible labeling
$(y_1, \dots, y_n) \in \{0,1\}^n$, there exists a classifier
$h \in \mathcal{H}$ such that:

$$
h(x_i) = y_i \quad \text{for all } i = 1, \dots, n
$$

---
## 2. VC Dimension

The **VC dimension** of a hypothesis class $\mathcal{H}$ is defined as:

$$
\text{VC}(\mathcal{H}) = \max \left\{ n : \exists \{x_1, \dots, x_n\}
\text{ shattered by } \mathcal{H} \right\}
$$

If arbitrarily large sets can be shattered, the VC dimension is infinite.

---
## 3. Intuition

- VC dimension measures how many points a classifier can label in
  **all possible ways**.
- Larger VC dimension means a more expressive (more complex) model.
- Small VC dimension implies stronger generalization guarantees.

In simple terms:
> VC dimension is the largest number of points that a model can
> classify correctly under **any labeling**.

---
## 4. Example: What does VC dimension = 3 mean?

Saying that a classifier has **VC dimension 3** means:

- There exists a set of 3 points that the classifier can shatter.
- For these 3 points, **all $2^3 = 8$ possible labelings** can be realized
  by some classifier in the class.
- However, **no set of 4 points** can be shattered.

In other words:

- The model is flexible enough to memorize any labeling of 3 points.
- But it cannot memorize all possible labelings of 4 points.

---
## 5. Concrete examples

- Threshold classifiers on the real line:
  VC dimension = 1

- Intervals on the real line:
  VC dimension = 2

- Linear classifiers in $\mathbb{R}^2$ (lines in the plane):
  VC dimension = 3

- Linear classifiers in $\mathbb{R}^d$:
  VC dimension = $d + 1$

---
## 6. Why VC dimension matters

VC dimension appears in generalization bounds.
With probability at least $1 - \delta$:

$$
R(h) \le \hat R(h) +
O\left( \sqrt{\frac{\text{VC}(\mathcal{H}) + \log(1/\delta)}{n}} \right)
$$

where:
- $R(h)$ is the true risk
- $\hat R(h)$ is the empirical risk
- $n$ is the sample size

---
## 7. Interpretation for learning

- Low VC dimension:
  underfitting risk, but strong generalization

- High VC dimension:
  expressive models, but risk of overfitting

VC dimension formalizes the **bias–variance tradeoff**.

---
## 8. Summary

- VC dimension measures the expressive power of a hypothesis class.
- VC dimension = 3 means:
  "any labeling of 3 points is possible, but not of 4".
- It is a fundamental concept in statistical learning theory.

# Random Projection & Johnson-Lindenstrauss Lemma

## 1. Wprowadzenie
**Random Projection (RP)** to technika redukcji wymiarowości, która opiera się na zaskakującej właściwości geometrii wysokich wymiarów. Pozwala ona na rzutowanie danych z bardzo wysokiego wymiaru $d$ do znacznie niższego wymiaru $k$, przy jednoczesnym zachowaniu odległości euklidesowych między punktami z minimalnym błędem.

Podstawą teoretyczną tej metody jest **Lemat Johnsona-Lindenstraussa (JL Lemma)**.

---

## 2. Lemat Johnsona-Lindenstraussa
Lemat ten mówi, że jeśli mamy $n$ punktów w przestrzeni o dowolnie wysokim wymiarze, to możemy je rzucić na przestrzeń o wymiarze $k \approx \log n$, a wzajemne odległości między punktami zostaną zachowane z dokładnością do $\epsilon$.



### Sformułowanie matematyczne:
Dla dowolnego zbioru $n$ punktów $V$ w $\mathbb{R}^d$ i parametru błędu $0 < \epsilon < 1$, istnieje odwzorowanie liniowe $f: \mathbb{R}^d \to \mathbb{R}^k$ dla wymiaru:
$$k \geq \frac{8 \ln n}{\epsilon^2}$$
takie, że dla każdych dwóch punktów $u, v \in V$ zachodzi:
$$(1 - \epsilon) \|u - v\|^2 \leq \|f(u) - f(v)\|^2 \leq (1 + \epsilon) \|u - v\|^2$$

**Kluczowy wniosek:** Docelowy wymiar $k$ zależy logarytmicznie od liczby punktów ($n$), a **nie zależy** od oryginalnego wymiaru ($d$).

---

## 3. Związek z Gaussian Annulus Theorem
Dlaczego losowe rzutowanie działa? Odpowiedź leży w **Gaussian Annulus Theorem**:
1.  W wysokich wymiarach masa rozkładu Gaussa skupia się w cienkim pierścieniu wokół sfery o promieniu $\sqrt{n}$.
2.  Dowolne dwa losowe wektory w wysokim wymiarze są z ogromnym prawdopodobieństwem **niemal prostopadłe** (ortogonalne).
3.  Dzięki temu losowa macierz rzutowania zachowuje strukturę geometryczną danych, działając jak "obiektyw", który nie zniekształca proporcji.



---

## 4. Porównanie: Random Projection vs. PCA
| Cecha | Random Projection | PCA |
| :--- | :--- | :--- |
| **Złożoność obliczeniowa** | $O(dk n)$ (bardzo szybka) | $O(d^2 n + d^3)$ (wolna dla dużych $d$) |
| **Zależność od danych** | Brak (macierz jest losowa) | Wysoka (wymaga analizy kowariancji) |
| **Cel** | Zachowanie odległości między punktami | Maksymalizacja wyjaśnionej wariancji |
| **Zastosowanie** | Bardzo duże zbiory danych, streaming | Eksploracja danych, wizualizacja |

---

```python
from sklearn.random_projection import GaussianRandomProjection
import numpy as np

# Przykładowe dane: 1000 próbek, 10 000 cech
X = np.random.rand(1000, 10000)

# Redukcja wymiarowości z zachowaniem błędu epsilon=0.1
transformer = GaussianRandomProjection(eps=0.1)
X_reduced = transformer.fit_transform(X)

print(f"Oryginalny wymiar: {X.shape[1]}")
print(f"Zredukowany wymiar: {X_reduced.shape[1]}")

# Estymacja Największej Wiarygodności (MLE) dla Macierzy Przejścia Łańcucha Markowa

## 1. Problem Estymacji
Mając daną zaobserwowaną sekwencję stanów $X = (x_0, x_1, \dots, x_T)$ wygenerowaną przez proces Markowa o $S$ stanach, naszym celem jest wyznaczenie macierzy przejścia $P$, gdzie każdy element $P_{ij}$ określa prawdopodobieństwo:
$$P_{ij} = P(X_{t+1} = j \mid X_t = i)$$

## 2. Statystyki Dostateczne: Macierz Liczników
Kluczową informacją potrzebną do estymacji są liczby przejść między poszczególnymi stanami. Definiujemy $n_{ij}$ jako liczbę wystąpień sekwencji $(i, j)$ w danych:
$$n_{ij} = \sum_{t=0}^{T-1} \mathbb{1}(x_t = i, x_{t+1} = j)$$



## 3. Funkcja Wiarygodności (Likelihood)
Własność Markowa mówi, że prawdopodobieństwo całej sekwencji jest iloczynem prawdopodobieństw kolejnych kroków. Funkcja wiarygodności $L(P)$ dla macierzy $P$ przy danych $X$ wynosi:
$$L(P) = P(x_0) \prod_{t=0}^{T-1} P_{x_t, x_{t+1}} = P(x_0) \prod_{i=1}^{S} \prod_{j=1}^{S} P_{ij}^{n_{ij}}$$

Przechodząc na logarytm (Log-Likelihood), otrzymujemy:
$$\ell(P) = \ln P(x_0) + \sum_{i=1}^{S} \sum_{j=1}^{S} n_{ij} \ln(P_{ij})$$

## 4. Wyprowadzenie Estymatora
Musimy zmaksymalizować $\ell(P)$ przy zachowaniu warunku, że każdy wiersz macierzy $P$ musi być rozkładem prawdopodobieństwa:
$$\sum_{j=1}^{S} P_{ij} = 1 \quad \forall i$$

Stosując metodę mnożników Lagrange'a dla każdego wiersza, otrzymujemy rozwiązanie:
$$\hat{P}_{ij}^{MLE} = \frac{n_{ij}}{\sum_{k=1}^{S} n_{ik}}$$



**Intuicja:** Prawdopodobieństwo przejścia z $i$ do $j$ to liczba zaobserwowanych przejść $i \to j$ podzielona przez łączną liczbę wyjść ze stanu $i$.

---

## 5. Implementacja w Python (NumPy)

```python
import numpy as np

def get_mle_transition_matrix(sequence, n_states):
    """
    Oblicza macierz przejść metodą MLE.
    """
    # 1. Inicjalizacja macierzy liczników n_ij
    counts = np.zeros((n_states, n_states))
    
    # 2. Zliczanie przejść w sekwencji
    for (i, j) in zip(sequence[:-1], sequence[1:]):
        counts[i, j] += 1
        
    # 3. Normalizacja wierszy (suma każdego wiersza = 1)
    row_sums = counts.sum(axis=1, keepdims=True)
    
    # Obsługa stanów, które nie wystąpiły (uniknięcie dzielenia przez 0)
    P_hat = np.divide(counts, row_sums, out=np.zeros_like(counts), where=row_sums!=0)
    
    return P_hat

# Przykład użycia:
# 0: Zimno, 1: Ciepło
dane = [0, 1, 1, 0, 1, 1, 1, 0, 0]
P = get_mle_transition_matrix(dane, n_states=2)
print(P)

# Wymiar VC (Vapnik-Chervonenkis Dimension)

## 1. Wstęp
**Wymiar VC** to miara złożoności (pojemności) rodziny modeli klasyfikacji binarnej. Określa on zdolność modelu do dopasowania się do dowolnie etykietowanych danych. Jest to kluczowe pojęcie w statystycznej teorii uczenia się, pozwalające oszacować błąd generalizacji.

## 2. Pojęcie Rozbijania (Shattering)
Zanim zdefiniujemy wymiar VC, musimy zdefiniować **rozbijanie**. 

Zbiór $n$ punktów jest **rozbijany** przez rodzinę modeli $H$, jeśli dla każdej z $2^n$ możliwych kombinacji etykiet (+1/-1) istnieje w tej rodzinie taki model, który poprawnie klasyfikuje wszystkie punkty.



## 3. Definicja Wymiaru VC
**Wymiar VC** klasy modeli $H$, oznaczany jako $d_{VC}(H)$, to wielkość **największego** zbioru punktów, który może być rozbity przez tę klasę.

* Jeśli wymiar VC wynosi $k$, oznacza to, że istnieje przynajmniej jeden układ $k$ punktów, który da się rozbić.
* Jednocześnie **żaden** układ $k+1$ punktów nie może zostać rozbity przez tę klasę.

### Przykład: Klasyfikator liniowy w 2D (prosta)
1.  **3 punkty:** Można je rozbić (o ile nie leżą na jednej linii). Istnieje 8 kombinacji etykiet i każdą da się oddzielić prostą.
2.  **4 punkty:** Nie da się ich rozbić. Jeśli ułożymy je w czworokąt i nadamy etykiety naprzemiennie (problem XOR), prosta linia nie oddzieli klas.
3.  **Wniosek:** Dla prostych w 2D, $d_{VC} = 3$.

Dla hiperpłaszczyzn w przestrzeni $d$-wymiarowej: 
$$d_{VC} = d + 1$$

## 4. Teoria VC i Generalizacja
Wymiar VC pozwala powiązać błąd treningowy ($E_{in}$) z błędem na nowych danych ($E_{out}$). Zgodnie z teorią Vapnika-Chervonenkisa, z prawdopodobieństwem co najmniej $1-\delta$ zachodzi:

$$E_{out} \leq E_{in} + \sqrt{\frac{8}{n} \ln \left( \frac{4(2n)^{d_{VC}}}{\delta} \right)}$$

Gdzie:
* $E_{in}$ – błąd na zbiorze treningowym.
* $n$ – liczba próbek.
* $d_{VC}$ – wymiar VC modelu.



## 5. Praktyczne znaczenie
* **Underfitting:** Model o zbyt niskim $d_{VC}$ nie jest w stanie nauczyć się wzorców (wysokie $E_{in}$).
* **Overfitting:** Model o bardzo wysokim $d_{VC}$ (np. głębokie sieci neuronowe, drzewa bez przycinania) może rozbić prawie każdy zbiór, co prowadzi do zapamiętywania szumu.
* **Zasada kciuka:** Aby uzyskać dobrą generalizację, liczba danych treningowych $n$ powinna być rzędu $10 \cdot d_{VC}$.

---
